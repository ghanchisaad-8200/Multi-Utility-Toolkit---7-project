```python
import datetime
import math
import random
import uuid
import os
import string
import secrets
```

```python
def show_current_datetime():
    print("Current Date and Time:", datetime.datetime.now())

def show_calendar():
    year = int(input("Enter year: "))
    month = int(input("Enter month (1-12): "))
    import calendar
    print(calendar.month(year, month))

def calculate_age():
    birth_year = int(input("Enter your birth year: "))
    current_year = datetime.datetime.now().year
    print("Your age is:", current_year - birth_year)
```

```python
def factorial():
    num = int(input("Enter a number: "))
    if num < 0:
        print("Factorial is not possible for negative numbers.")
    else:
        print("Factorial:", math.factorial(num))

def compound_interest():
    principal = float(input("Enter principal amount: "))
    rate = float(input("Enter annual interest rate (%): "))
    time = float(input("Enter time in years: "))

    amount = principal * (1 + rate / 100) ** time
    print("Compound Interest:", round(amount - principal, 2))
    print("Total Amount:", round(amount, 2))

def circle_area():
    radius = float(input("Enter circle radius: "))
    print("Area of Circle:", math.pi * radius ** 2)

def square_root():
    num = float(input("Enter a number: "))
    if num < 0:
        print("Square root of a negative number is not supported.")
    else:
        print("Square Root:", math.sqrt(num))
```

```python
def random_number():
    start = int(input("Enter starting number: "))
    end = int(input("Enter ending number: "))
    print("Random Number:", random.randint(start, end))

def random_choice():
    items = input("Enter items separated by commas: ").split(",")
    items = [item.strip() for item in items if item.strip()]

    if items:
        print("Random Choice:", random.choice(items))
    else:
        print("No items entered.")

def generate_password():
    length = int(input("Enter password length: "))

    if length < 1:
        print("Password length must be at least 1.")
        return

    characters = string.ascii_letters + string.digits + string.punctuation
    password = ''.join(secrets.choice(characters) for _ in range(length))
    print("Generated Password:", password)

def generate_otp():
    otp = ''.join(secrets.choice(string.digits) for _ in range(6))
    print("Your OTP is:", otp)
```

```python
def generate_uuid():
    unique_id = uuid.uuid4()
    print("Generated UUID:", unique_id)

def generate_multiple_uuids():
    count = int(input("How many UUIDs do you want? "))

    if count < 1:
        print("Enter a number greater than zero.")
        return

    for i in range(count):
        print(i + 1, ":", uuid.uuid4())
```

```python
def create_file():
    filename = input("Enter file name: ")

    with open(filename, "w") as file:
        file.write("This file was created using Python.")

    print("File created successfully!")

def write_file():
    filename = input("Enter file name: ")
    content = input("Enter content: ")

    with open(filename, "w") as file:
        file.write(content)

    print("Content written successfully!")

def read_file():
    filename = input("Enter file name: ")

    try:
        with open(filename, "r") as file:
            print("\nFile Content:")
            print(file.read())
    except FileNotFoundError:
        print("File not found!")

def append_file():
    filename = input("Enter file name: ")
    content = input("Enter content to append: ")

    with open(filename, "a") as file:
        file.write("\n" + content)

    print("Content added successfully!")
```

```python
def explore_module():
    print("Available modules:")
    print("1. math")
    print("2. random")
    print("3. datetime")
    print("4. os")
    print("5. uuid")

    choice = input("Select a module: ")

    modules = {
        "1": math,
        "2": random,
        "3": datetime,
        "4": os,
        "5": uuid
    }

    if choice in modules:
        module = modules[choice]
        print("\nModule Name:", module.__name__)
        print("Available Functions and Attributes:")
        print(dir(module))
    else:
        print("Invalid choice!")
```

```python
def main():
    while True:
        print("\n========== MULTI-UTILITY TOOLKIT ==========")
        print("1. Current Date and Time")
        print("2. Show Calendar")
        print("3. Calculate Age")
        print("4. Factorial")
        print("5. Compound Interest")
        print("6. Circle Area")
        print("7. Square Root")
        print("8. Random Number")
        print("9. Random Choice")
        print("10. Generate Password")
        print("11. Generate OTP")
        print("12. Generate UUID")
        print("13. Generate Multiple UUIDs")
        print("14. Create File")
        print("15. Write File")
        print("16. Read File")
        print("17. Append File")
        print("18. Explore Python Modules")
        print("0. Exit")

        choice = input("\nEnter your choice: ")

        if choice == "1":
            show_current_datetime()
        elif choice == "2":
            show_calendar()
        elif choice == "3":
            calculate_age()
        elif choice == "4":
            factorial()
        elif choice == "5":
            compound_interest()
        elif choice == "6":
            circle_area()
        elif choice == "7":
            square_root()
        elif choice == "8":
            random_number()
        elif choice == "9":
            random_choice()
        elif choice == "10":
            generate_password()
        elif choice == "11":
            generate_otp()
        elif choice == "12":
            generate_uuid()
        elif choice == "13":
            generate_multiple_uuids()
        elif choice == "14":
            create_file()
        elif choice == "15":
            write_file()
        elif choice == "16":
            read_file()
        elif choice == "17":
            append_file()
        elif choice == "18":
            explore_module()
        elif choice == "0":
            print("Thank you for using Multi-Utility Toolkit!")
            break
        else:
            print("Invalid choice! Please try again.")
```
